```python
data_p2='/mnt/GSHD_DATA_01/partner/icool/revenue/data/DT_20220101_20220217.csv'
dt_2021=pd.read_csv(data_p2, sep='\t')
```

```python
dt_2021=dt_2021[['date', 'cuahang_id', 'total']]
dt_2021.columns=['ngay', 'macuahang', 'doanhthu']
```

```python
dt_2021.head()
```

```python
dt_2021['doanhthu']=round(dt_2021['doanhthu']/1000000,2)
```

```python
dt_2021.head()
```

```python
list_CH=list(set(dt_2021['macuahang'].tolist()))
list_CH=list_CH[:-2]
```

```python
list_CH
```

```python
for x in list_CH:
    ## read predict from 2022/02/18-2022/12/31
    data_p1=f'/mnt/GSHD_DATA_01/partner/icool/revenue/output/{x}/2022.02.18/predict_316.csv'
    data_01=pd.read_csv(data_p1)

    ## read gt from 2022/01/01-2022/02/17
    data_02=dt_2021[dt_2021['macuahang']==x]
    total=data_01['doanhthu'].tolist()+data_02['doanhthu'].tolist()
    total=round(sum(total)/1000, 4)

    print(f'==> doanh thu du bao cua cua hang: {x} la: {total} ty VND')
```
